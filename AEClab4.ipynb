Program 3: Continuous Distributions — Normal & Exponential


```python
import numpy as np #NumPy for numerical calculations and random data generation,
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats #SciPy for probability distributions
import pandas as pd

np.random.seed(42) # np.random.seed(42) makes our random experiment reproducible
```

SECTION A: NORMAL DISTRIBUTION

```python
print("=" * 60)
print("NORMAL DISTRIBUTION  N(μ, σ²)")
print("=" * 60)

params = [(0, 1), (2, 1), (0, 2), (-2, 0.5)]
x = np.linspace(-8, 8, 400)

print("\nNormal Distribution Properties:")
for mu, sigma in params:
    dist = stats.norm(mu, sigma)
    print(f"\n  N(μ={mu}, σ={sigma}):")
    print(f"  Mean={dist.mean():.2f}, Var={dist.var():.2f},"
          f" Skew={dist.stats('s'):.2f},"
          f" Kurt={dist.stats('k'):.2f}")

# Standard Normal — 68-95-99.7 Rule
std_normal = stats.norm(0, 1)
print("\n68-95-99.7 Rule Verification:")
print(f"  P(-1 < Z < 1)  = {std_normal.cdf(1) - std_normal.cdf(-1):.6f} ≈ 68.27%")
print(f"  P(-2 < Z < 2)  = {std_normal.cdf(2) - std_normal.cdf(-2):.6f} ≈ 95.45%")
print(f"  P(-3 < Z < 3)  = {std_normal.cdf(3) - std_normal.cdf(-3):.6f} ≈ 99.73%")

# Z-score and Standardization
print("\nZ-score Examples:")
raw_scores = [60, 70, 80, 90]
mu_exam, sigma_exam = 70, 10
for score in raw_scores:
    z = (score - mu_exam) / sigma_exam
    percentile = std_normal.cdf(z) * 100
    print(f"  Score={score}, Z={z:.2f}, Percentile={percentile:.1f}%")

# ML Context: Weight initialization
print("\n  ML Context — Weight Initialization:")
print(f"  Xavier init: W ~ N(0, 2/(n_in + n_out))")
n_in, n_out = 784, 128
xavier_std = np.sqrt(2 / (n_in + n_out))
print(f"  For Dense(784→128): σ = {xavier_std:.6f}")
```

SECTION B: EXPONENTIAL DISTRIBUTION

```python
print("\n" + "=" * 60)
print("EXPONENTIAL DISTRIBUTION  Exp(λ)")
print("=" * 60)

lambdas_exp = [0.5, 1.0, 2.0, 3.0]
x_exp = np.linspace(0, 8, 300)

for lam in lambdas_exp:
    dist_e = stats.expon(scale=1/lam)
    print(f"\n  Exp(λ={lam}): Mean={dist_e.mean():.4f},"
          f" Var={dist_e.var():.4f},"
          f" Std={dist_e.std():.4f}")

lam_main = 1.0
exp_main = stats.expon(scale=1/lam_main)
exp_data = np.random.exponential(1/lam_main, 5000)
print(f"\n  Memoryless Property (λ={lam_main}):")
print(f"  P(X > 2)            = {1 - exp_main.cdf(2):.6f}")
print(f"  P(X > 3 | X > 1)    = {(1 - exp_main.cdf(3)) / (1 - exp_main.cdf(1)):.6f}")
print(f"  Both equal? {abs((1-exp_main.cdf(2)) - (1-exp_main.cdf(3))/(1-exp_main.cdf(1))) < 1e-9}")
```

SECTION C: CENTRAL LIMIT THEOREM

```python
print("\n" + "=" * 60)
print("CENTRAL LIMIT THEOREM SIMULATION")
print("=" * 60)

distributions_clt = {
    'Uniform(0,1)': lambda n: np.random.uniform(0, 1, n),
    'Exponential(1)': lambda n: np.random.exponential(1, n),
    'Binomial(10,0.3)': lambda n: np.random.binomial(10, 0.3, n)
}

sample_sizes = [1, 5, 30, 100]
for dist_name, sampler in distributions_clt.items():
    print(f"\n  {dist_name}:")
    for n in sample_sizes:
        sample_means = [np.mean(sampler(n)) for _ in range(2000)]
        print(f"    n={n:>3}: mean={np.mean(sample_means):.4f},"
              f" std={np.std(sample_means):.4f}")
```

MOMENTS ANALYSIS

```python
print("\n" + "=" * 60)
print("MOMENTS ANALYSIS")
print("=" * 60)

print("\nRaw Moments E[Xᵏ]:")
dist_norm = stats.norm(0, 1)
for k in range(1, 5):
    raw_moment = dist_norm.moment(k)
    print(f"  E[X^{k}] = {raw_moment:.6f}")

print("\nCentral Moments:")
data_sample = np.random.normal(5, 2, 1000)
for k in range(1, 5):
    central = np.mean((data_sample - np.mean(data_sample))**k)
    print(f"  μ_{k} = {central:.6f}")
```

VISUALISATION

```python
fig, axes = plt.subplots(3, 3, figsize=(18, 15))
fig.suptitle('Continuous Distributions — Normal, Exponential & CLT',
             fontsize=16, fontweight='bold')

# 1. Normal PDFs
for mu, sigma in params:
    axes[0, 0].plot(x, stats.norm.pdf(x, mu, sigma),
                    linewidth=2, label=f'N({mu},{sigma})')
axes[0, 0].set_title('Normal Distribution PDFs')
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('f(x)')
axes[0, 0].legend(fontsize=8)
axes[0, 0].axhline(0, color='black', linewidth=0.5)

# 2. 68-95-99.7 Rule
x_std = np.linspace(-4, 4, 300)
y_std = stats.norm.pdf(x_std)
axes[0, 1].plot(x_std, y_std, 'k-', linewidth=2)
for interval, color, label in [
        (1, 'lightblue', '68.27%'),
        (2, 'lightyellow', '95.45%'),
        (3, 'lightcoral', '99.73%')]:
    axes[0, 1].fill_between(
        x_std, y_std,
        where=(np.abs(x_std) <= interval),
        alpha=0.4, color=color, label=f'±{interval}σ: {label}')
axes[0, 1].set_title('68-95-99.7 Rule')
axes[0, 1].legend(fontsize=8)
axes[0, 1].set_xlabel('Z (Standard Units)')

# 3. Normal CDF
x_cdf = np.linspace(-4, 4, 300)
axes[0, 2].plot(x_cdf, stats.norm.cdf(x_cdf, 0, 1),
                'blue', linewidth=2)
for q_val in [0.25, 0.50, 0.75, 0.95]:
    q_x = stats.norm.ppf(q_val)
    axes[0, 2].plot(q_x, q_val, 'ro', markersize=7)
    axes[0, 2].annotate(f'{q_val:.0%}',
                        xy=(q_x, q_val),
                        xytext=(q_x+0.3, q_val-0.05),
                        fontsize=8)
axes[0, 2].set_title('Normal CDF with Quantiles')
axes[0, 2].set_xlabel('x')
axes[0, 2].set_ylabel('F(x) = P(X ≤ x)')

# 4. Exponential PDFs
for lam in lambdas_exp:
    axes[1, 0].plot(x_exp, stats.expon.pdf(x_exp, scale=1/lam),
                    linewidth=2, label=f'λ={lam}')
axes[1, 0].set_title('Exponential Distribution PDFs')
axes[1, 0].set_xlabel('x')
axes[1, 0].legend()

# 5. Exponential — Memoryless Property
x_mem = np.linspace(0, 5, 200)
axes[1, 1].fill_between(x_mem, exp_main.pdf(x_mem),
                         where=(x_mem >= 1),
                         alpha=0.5, color='orange',
                         label='P(X>1)')
axes[1, 1].fill_between(x_mem, exp_main.pdf(x_mem),
                         where=(x_mem >= 2),
                         alpha=0.7, color='red',
                         label='P(X>2)=P(X>3|X>1)')
axes[1, 1].plot(x_mem, exp_main.pdf(x_mem), 'b-', linewidth=2)
axes[1, 1].set_title('Exponential — Memoryless Property')
axes[1, 1].legend()
axes[1, 1].set_xlabel('x')

# 6. CLT for different sample sizes
clt_data = np.random.exponential(1, (100, 10000))
for i, n in enumerate([1, 5, 30, 100]):
    means = clt_data[:n, :].mean(axis=0)
    axes[1, 2].hist(means, bins=60, density=True, alpha=0.5,
                    label=f'n={n}')
axes[1, 2].set_title('CLT: Sample Means of Exp(1)')
axes[1, 2].set_xlabel('Sample Mean')
axes[1, 2].legend()

# 7-9: CLT for different distributions
for idx, (dist_name, sampler) in enumerate(
        distributions_clt.items()):
    ax = axes[2, idx]
    for n in [1, 5, 30]:
        sample_means = [np.mean(sampler(n)) for _ in range(2000)]
        ax.hist(sample_means, bins=50, density=True,
                alpha=0.5, label=f'n={n}')
    ax.set_title(f'CLT: {dist_name}')
    ax.set_xlabel('Sample Mean')
    ax.legend(fontsize=8)

plt.tight_layout()
plt.savefig('P3_continuous_distributions.png', dpi=150,
            bbox_inches='tight')
plt.show()
print("\nProgram 3 Completed Successfully.")
```

EXAMPLE CASE STUDY:
Adult Blood Pressure
Systolic: μ ≈ 120 mmHg, σ ≈ 14 mmHg
Systolic blood pressure in a general adult population follows approximately a normal distribution. The American Heart Association sets the Stage 1 hypertension threshold at ≥130 mmHg — a z-score of about 0.71 on the population distribution, meaning roughly 24% of adults meet or exceed that threshold. Clinical screening programs use this to size testing needs.

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Define the population parameters
mu = 120         # Mean systolic blood pressure (mmHg)
sigma = 14       # Standard deviation (mmHg)
threshold = 130  # Stage 1 hypertension threshold (mmHg)

# Calculate Z-score and probability
z_score = (threshold - mu) / sigma
probability = stats.norm.sf(threshold, loc=mu, scale=sigma)
percentage = probability * 100

# --- GRAPHING SECTION ---

# Generate data for the x-axis (4 standard deviations on either side of the mean)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

# Calculate the y-values (probability density function)
y = stats.norm.pdf(x, mu, sigma)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='dodgerblue', linewidth=2, label='Population Distribution')

# Shade the area under the curve for X >= 130
x_fill = np.linspace(threshold, mu + 4*sigma, 500)
y_fill = stats.norm.pdf(x_fill, mu, sigma)
plt.fill_between(x_fill, y_fill, color='salmon', alpha=0.5,
                 label=f'Stage 1 Hypertension\n(>=130 mmHg, {percentage:.1f}%)')

# Add vertical lines for the mean and the threshold
plt.axvline(mu, color='green', linestyle='--', linewidth=1.5, label=f'Mean (μ) = {mu}')
plt.axvline(threshold, color='red', linestyle='--', linewidth=1.5, label=f'Threshold = {threshold}')

# Add text annotation for the Z-score
plt.text(threshold + 2, 0.015, f'Z-score = {z_score:.2f}', fontsize=12, color='darkred')

# Formatting the graph
plt.title('Adult Systolic Blood Pressure Distribution', fontsize=14, pad=15)
plt.xlabel('Systolic Blood Pressure (mmHg)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.legend(loc='upper right', frameon=True, shadow=True)
plt.grid(axis='y', alpha=0.3)
plt.xlim(mu - 4*sigma, mu + 4*sigma)
plt.tight_layout()

# Display the plot
plt.show()

# Print the text results
print(f"\nPopulation Mean (μ): {mu}")
print(f"Standard Deviation (σ): {sigma}")
print(f"Threshold (X): {threshold}\n")

print(f"Calculated Z-score: {z_score:.4f}")
print(f"Proportion of population >= 130 mmHg: {probability:.4f}")
print(f"Percentage of adults: {percentage:.2f}%")
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Set up the 3x3 grid
fig, axes = plt.subplots(3, 3, figsize=(18, 15))
fig.suptitle('Comprehensive Statistical Analysis: Blood Pressure & Clinic Operations',
             fontsize=20, fontweight='bold', y=0.98)

# --- Row 1: The Individual Normal Distribution ---
mu_bp = 120
sigma_bp = 14
x_norm = np.linspace(mu_bp - 4*sigma_bp, mu_bp + 4*sigma_bp, 1000)

# 1. Normal PDF
ax = axes[0, 0]
y_norm = stats.norm.pdf(x_norm, mu_bp, sigma_bp)
ax.plot(x_norm, y_norm, color='dodgerblue', lw=2)
ax.fill_between(x_norm[x_norm >= 130], y_norm[x_norm >= 130], color='salmon', alpha=0.6, label='>= 130 mmHg (24%)')
ax.axvline(120, color='green', linestyle='--', label='Mean (120)')
ax.axvline(130, color='red', linestyle='--', label='Threshold (130)')
ax.set_title('1. Population BP (PDF)', fontsize=14)
ax.set_xlabel('Systolic BP (mmHg)')
ax.set_ylabel('Probability Density')
ax.legend()
ax.grid(alpha=0.3)

# 2. Normal CDF
ax = axes[0, 1]
y_cdf = stats.norm.cdf(x_norm, mu_bp, sigma_bp)
prob_130 = stats.norm.cdf(130, mu_bp, sigma_bp)
ax.plot(x_norm, y_cdf, color='darkorange', lw=2)
ax.axhline(prob_130, color='red', linestyle='--', alpha=0.7, label=f'{prob_130*100:.1f}th Percentile')
ax.axvline(130, color='red', linestyle='--', alpha=0.7)
ax.set_title('2. Cumulative Probability (CDF)', fontsize=14)
ax.set_xlabel('Systolic BP (mmHg)')
ax.set_ylabel('Cumulative Probability')
ax.legend()
ax.grid(alpha=0.3)

# 3. Z-Score (Standard Normal)
ax = axes[0, 2]
z_x = np.linspace(-4, 4, 1000)
z_y = stats.norm.pdf(z_x, 0, 1)
z_val = (130 - 120) / 14
ax.plot(z_x, z_y, color='purple', lw=2)
ax.fill_between(z_x[z_x >= z_val], z_y[z_x >= z_val], color='plum', alpha=0.6, label=f'Z >= {z_val:.2f}')
ax.axvline(0, color='black', linestyle='--', label='Mean (Z=0)')
ax.axvline(z_val, color='red', linestyle='--')
ax.set_title('3. Standardized Z-Score (PDF)', fontsize=14)
ax.set_xlabel('Standard Deviations (Z)')
ax.legend()
ax.grid(alpha=0.3)

# --- Row 2: The Exponential Distribution (Clinic Operations) ---
lambda_rate = 10 # 10 patients per hour
x_exp = np.linspace(0, 0.5, 1000) # Up to 30 mins (0.5 hours)

# 4. Exponential PDF
ax = axes[1, 0]
y_exp = stats.expon.pdf(x_exp, scale=1/lambda_rate)
ax.plot(x_exp, y_exp, color='teal', lw=2)
ax.fill_between(x_exp, y_exp, color='paleturquoise', alpha=0.5)
ax.axvline(1/lambda_rate, color='darkslategray', linestyle='--', label=f'Mean Wait ({60/lambda_rate:.0f} mins)')
ax.set_title('4. Time Between Arrivals (PDF)', fontsize=14)
ax.set_xlabel('Time (Hours)')
ax.set_ylabel('Density')
ax.legend()
ax.grid(alpha=0.3)

# 5. Exponential CDF
ax = axes[1, 1]
y_exp_cdf = stats.expon.cdf(x_exp, scale=1/lambda_rate)
ax.plot(x_exp, y_exp_cdf, color='mediumseagreen', lw=2)
ax.set_title('5. Prob. Arrival Within X Hours (CDF)', fontsize=14)
ax.set_xlabel('Time (Hours)')
ax.set_ylabel('Cumulative Probability')
ax.grid(alpha=0.3)

# 6. Exponential Survival Function
ax = axes[1, 2]
y_exp_sf = stats.expon.sf(x_exp, scale=1/lambda_rate)
ax.plot(x_exp, y_exp_sf, color='indianred', lw=2)
ax.set_title('6. Prob. Waiting MORE than X Hours', fontsize=14)
ax.set_xlabel('Time (Hours)')
ax.set_ylabel('Survival Probability')
ax.grid(alpha=0.3)

# --- Row 3: Central Limit Theorem (Sample Sizes) ---
# 7. CLT n = 5
ax = axes[2, 0]
n1 = 5
se1 = sigma_bp / np.sqrt(n1)
y_clt1 = stats.norm.pdf(x_norm, mu_bp, se1)
ax.plot(x_norm, y_norm, color='lightgray', linestyle='--', label='Individual BP')
ax.plot(x_norm, y_clt1, color='blue', lw=2, label=f'Means (n={n1})')
ax.fill_between(x_norm, y_clt1, color='lightblue', alpha=0.4)
ax.set_title('7. CLT: Small Clinic Sample (n=5)', fontsize=14)
ax.set_xlabel('Average BP (mmHg)')
ax.legend()
ax.set_xlim(80, 160)
ax.grid(alpha=0.3)

# 8. CLT n = 30
ax = axes[2, 1]
n2 = 30
se2 = sigma_bp / np.sqrt(n2)
y_clt2 = stats.norm.pdf(x_norm, mu_bp, se2)
ax.plot(x_norm, y_norm, color='lightgray', linestyle='--')
ax.plot(x_norm, y_clt2, color='darkorange', lw=2, label=f'Means (n={n2})')
ax.fill_between(x_norm, y_clt2, color='moccasin', alpha=0.5)
ax.set_title('8. CLT: Standard Study Size (n=30)', fontsize=14)
ax.set_xlabel('Average BP (mmHg)')
ax.legend()
ax.set_xlim(80, 160)
ax.grid(alpha=0.3)

# 9. CLT n = 100
ax = axes[2, 2]
n3 = 100
se3 = sigma_bp / np.sqrt(n3)
y_clt3 = stats.norm.pdf(x_norm, mu_bp, se3)
ax.plot(x_norm, y_norm, color='lightgray', linestyle='--')
ax.plot(x_norm, y_clt3, color='forestgreen', lw=2, label=f'Means (n={n3})')
ax.fill_between(x_norm, y_clt3, color='lightgreen', alpha=0.5)
ax.set_title('9. CLT: Large Hospital Sample (n=100)', fontsize=14)
ax.set_xlabel('Average BP (mmHg)')
ax.legend()
ax.set_xlim(80, 160)
ax.grid(alpha=0.3)

# Adjust layout and display
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
