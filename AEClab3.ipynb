```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
np.random.seed(42)
n, p = 20, 0.35
binom = stats.binom(n,p)
x = np.arange(0,n+1)

print(f" Mean = {binom.mean() :.4f} (Theory: np = {n*p})")
print(f" Variance = {binom.var() :.4f}")
print(f" Second Moment = {binom.moment(2) :.4f}")
print(f" Third Moment = {binom.moment(3) :.4f}")
print(f" Fourth Moment = {binom.moment(4) :.4f}")
print(f" Skewness = {binom.stats('s') :.4f}")
print(f" CDF = {binom.cdf(10) :.4f}")
print(f" PMF = {binom.pmf(7) :.4f}")

# 1. Binomial PMF
plt.figure(figsize=(7, 5))
plt.bar(x, binom.pmf(x), color='steelblue', edgecolor='black', alpha=0.8)

plt.axvline(binom.mean(), color='red', linestyle='--', label=f'Mean = {binom.mean():.1f}')

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
for lam in [1, 3, 5, 10]:
  d = stats.poisson(lam)
  sim = np.random.poisson(lam, 5000)
  print(f"Mean = {d.mean() :.4f}")
  sim = np.random.poisson(lam, 5000)
  print(f"Simulation Mean = {sim.mean() :.4f}")

plt.figure(figsize=(7, 5))
x_p = np.arange(0, 20)
for lam in [1, 3, 5, 10]:
    plt.plot(x_p, stats.poisson.pmf(x_p, lam), marker='o', markersize=4, label=f'λ = {lam}')

plt.title('Poisson PMF — Effect of λ')
plt.xlabel('Number of Events (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
p_geo = 0.3
geo = stats.geom(p_geo)
print(f"Geo Mean = {geo.mean():.4f}")
print(f"Geo PMF = {geo.pmf(1):.4f}")
print(f"Geo CDF = {1 - geo.cdf(5):.4f}")

plt.figure(figsize=(7, 5))
x_g = np.arange(1, 20)

plt.bar(x_g, geo.pmf(x_g), color='green', edgecolor='black', alpha=0.8)

plt.title(f'Geometric PMF (p={p_geo})')
plt.xlabel('Trial Number (x)')
plt.ylabel('Probability')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
